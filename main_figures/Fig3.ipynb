# Fashion MNIST

```python
fashion_active_sim = np.load(savePath / 'fashion_active_sim.npy', allow_pickle=True)
fashion_active_exp = np.load(savePath / 'fashion_active_exp.npy', allow_pickle=True)
```

```python
common_kwg = {'capsize':0,'alpha':0.89,'ms':6}
k_val = 1

fig, ax = plt.subplots(figsize=(4.3,3.2),dpi=120)
plt.xlabel(r'Number of illumination patterns $d_{\mathrm{f}}$')
plt.ylabel('Test accuracy (%)')

exp_pans = fashion_active_exp[fashion_active_sim['K'] == 1].sort_values('N')
sim_pans = fashion_active_sim[fashion_active_exp['K'] == 1].sort_values('N')
nstd = 3 

ax.errorbar(exp_pans['N'],exp_pans['mean_accuracy'],yerr=exp_pans['std_dev'],fmt='o',c=c_red,label='Experiment: active PANS',**common_kwg,zorder=100)
ax.errorbar(exp_pans['N'],sim_pans['mean_accuracy'],fmt='.-',lw=1.8,color=c_red,alpha=0.15,label='Simulation: active PANS')
ax.fill_between(exp_pans['N'],(sim_pans['mean_accuracy']+sim_pans['std_dev']*nstd),(sim_pans['mean_accuracy']-sim_pans['std_dev']*nstd),color=c_red,alpha=0.08,lw=0)
ax.set_ylim(48,90.4)

plt.legend()
plt.tight_layout()
plt.savefig(plotPath / f'Fig3_fashion.png')
```
