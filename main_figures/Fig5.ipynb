# MMF classification

```python
MMF_MNIST_classification_passive_sim = np.load(savePath / 'MMF_MNIST_classification_passive_sim.npy', allow_pickle=True)
MMF_MNIST_classification_direct_imaging = np.load(savePath / 'MMF_MNIST_classification_directimg.npy', allow_pickle=True)
```

```python
common_kwg = {'capsize':0,'lw':2,'ms':5}
sx,sy = 4.2,2.7
fig, ax = plt.subplots(figsize=(sx,sy),dpi=120)
plt.xlabel(r'Total number of detected photons $N_{det}$')
plt.ylabel('Test accuracy (%)')

for i, dcr in enumerate(MMF_MNIST_classification_passive_sim['dcr'].unique()):
    the_df = MMF_MNIST_classification_passive_sim[MMF_MNIST_classification_passive_sim['dcr'] == dcr].sort_values('d_f')
    plt.errorbar(the_df['det_photon_count'],the_df['mean_accuracy'],yerr=the_df['std_dev'],fmt='-o',\
        c=c_red,ms=5,alpha=1-(i*0.45),lw=2,label=f'PANS: DCR {dcr*100:.0f}%') 

plt.errorbar(MMF_MNIST_classification_direct_imaging['det_photon_count'],MMF_MNIST_classification_direct_imaging['mean_accuracy'],yerr=MMF_MNIST_classification_direct_imaging['std_dev'],\
    fmt='.-',label=f'Direct imaging: ideal',**common_kwg)
plt.plot([0,27],[10,]*2,'--',c='gray',lw=2)
plt.xlim(0.7,24)
plt.yticks(np.arange(10,100,20))
plt.legend()
plt.tight_layout()
plt.savefig(plotPath / 'Fig5_MMF_classification.png')
```

```python
MMF_MNIST_classification_direct_imaging = np.load(savePath / 'MMF_MNIST_classification_directimg_high.npy', allow_pickle=True)

sx,sy = 2.,1.2
fig, ax = plt.subplots(figsize=(sx,sy),dpi=120)
# plt.xlabel(r'Total number of detected photons $N_{det}$')
# plt.ylabel('Test accuracy (%)')
ax.errorbar(MMF_MNIST_classification_direct_imaging['det_photon_count'],MMF_MNIST_classification_direct_imaging['mean_accuracy'],yerr=MMF_MNIST_classification_direct_imaging['std_dev'],\
            label='Direct imaging',fmt='-',**common_kwg)

ax.set_xticks(10**np.arange(5))
ax.set_xlim(1,1.8e4)
ax.set_xscale('log')
```
