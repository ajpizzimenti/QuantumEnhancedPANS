# Cell classification

```python
cell_active_sim = np.load(savePath / 'cell_active_sim.npy', allow_pickle=True)
cell_direct_imaging = np.load(savePath / 'cell_directimg.npy', allow_pickle=True)
```

```python
common_kwg = {'capsize':0,'lw':2,'ms':5}
sx,sy = 4.5,2.7
fig, ax = plt.subplots(figsize=(sx,sy),dpi=120)
plt.xlabel(r'Total number of detected photons $N_{det}$')
plt.ylabel('Test accuracy (%)')

for i, dcr in enumerate(cell_active_sim['dcr'].unique()):
    the_df = cell_active_sim[cell_active_sim['dcr'] == dcr].sort_values('d_f')
    plt.errorbar(the_df['det_photon_count'],the_df['mean_accuracy'],yerr=the_df['std_dev'],fmt='-o',\
        c=c_red,ms=5,alpha=1-(i*0.3),lw=2,label=f'PANS: DCR {dcr*100:.0f}%')


plt.errorbar(cell_direct_imaging['det_photon_count'],cell_direct_imaging['mean_accuracy'],yerr=cell_direct_imaging['std_dev'],\
    fmt='.-',label=f'Direct imaging: ideal',**common_kwg)
plt.plot([0,77],[100/3.,]*2,'--',c='gray',lw=2)
plt.xlim(0.7,21.7)
plt.legend()
plt.tight_layout()
plt.savefig(plotPath / 'Fig4_cell.png')
```
